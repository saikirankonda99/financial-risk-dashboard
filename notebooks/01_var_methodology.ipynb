# VaR Methodology Comparison
Compare Historical, Parametric, and Monte Carlo VaR on the same portfolio.

```python
import sys; sys.path.insert(0, '..')
import pandas as pd
from src.risk.var import historical_var, parametric_var, monte_carlo_var

returns = pd.read_csv('../data/sample_returns.csv')['return'].values
pv = pd.read_csv('../data/sample_portfolio.csv')['market_value'].sum()

for fn in (historical_var, parametric_var, monte_carlo_var):
    r = fn(returns, pv, confidence=0.99)
    print(f'{r.method:15s} VaR: ${r.var_dollar:12,.0f}  ES: ${r.es_dollar:12,.0f}')
```
